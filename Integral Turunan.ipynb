Metode Integral Numerik

```python
import numpy as np

# Mendefinisikan fungsi yang diintegrasikan
def f(x):
    return np.exp(-x**2)

# Mendefinisikan fungsi aturan trapesium
def trapesium(a, b, n):
    # Menentukan nilai h
    h = (b - a) / n

    # Menentukan nilai titik partisi
    x = np.linspace(a, b, n + 1)

    # Menentukan nilai fungsi untuk setiap titik partisi
    y = f(x)

    # Rumus aturan trapesium
    integral = (h / 2) * (y[0] + 2 * np.sum(y[1:-1]) + y[-1])

    return integral, h

# Menentukan batas integral
a = 0.2
b = 2.6

# Nilai acuan
I_acuan = 0.688653

# Jumlah sub interval/partisi
nilai_n = [6, 12, 18, 24]

# Menampilkan tabel hasil
print("Hasil Integrasi Menggunakan Aturan Trapesium")
print("_" * 70)
print(f"{'n':>5} {'h':>12} {'Integral':>18} {'Galat Absolut':>18}")
print("_" * 70)

# Perulangan untuk setiap nilai n
for n in nilai_n:
    hasil, h = trapesium(a, b, n)

    # Menghitung galat
    galat = abs(I_acuan - hasil)

    # Menampilkan hasil
    print(f"{n:>5} {h:>12.6f} {hasil:>18.8f} {galat:>18.8f}")

print("_" * 70)
```
