Studi Kasus: Estimasi Throughput Jaringan

```python
import numpy as np
# Pustaka baru scipy dengan fungsi lagrange
from scipy.interpolate import lagrange 

print("="*60)
print("IMPLEMENTASI POLINOMIAL INTERPOLASI LAGRANGE")
print("="*60)

# Input data mentah
x = np.array([20.0, 50.0, 80.0], dtype=float)
y = np.array([45.0, 80.0, 95.0], dtype=float)

# Menghitung Nilai Interpolasi
poly_lagrange = lagrange(x, y)

# Menghitung nilai y yang Dicari
x_target = 60
y_pred = poly_lagrange(x_target)

# Menampilkan Hasil
print(f"Nilai data x (%): {x}")
print(f"Nilai data y (Mbps): {y}")
print(f"Estimasi nilai throughput pada saat:{x_target} % adalah {y_pred:.2f} Mbps")
```
