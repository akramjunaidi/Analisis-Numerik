Studi Kasus: Ukuran File Model AI vs Akurasi

```python
import numpy as np

print("="*60)
print("IMPLEMENTASI POLINOMIAL APROKSIMASI: Least Square")
print("="*60)

# Input Data
x = np.array([10.0, 20.0, 30.0, 40.0], dtype=float)
y = np.array([82.0, 85.0, 88.0, 92.0], dtype=float)

# Menghitung Garis Aproksimasi y = ax + b
a, b = np.polyfit(x, y, 1)

# Menghitung Estimasi Nilai y yang Dicari
x_target = 25.0
y_pred = a * x_target + b

# Menampilkan Hasil
print(f"Nilai data x (MB): {x}")
print(f"Nilai data y (%): {y}")
print(f"Persamaan garis aproksimasi: y = {a:.2f}x + {b:.2f}")
print(f"Estimasi nilai akurasi model (y) saat x = {x_target} MB adalah {y_pred:.2f} %")
```
